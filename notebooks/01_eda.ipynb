```python
# ============================================================
# BLOCKING TEST - SMALL / FAST VERSION
# ============================================================

import re
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent


# ============================================================
# 1. LOAD SMALL SAMPLE
# ============================================================

print("Loading data...")

s1 = pd.read_csv(
    PROJECT_ROOT / "dataset/train/train_source1.tsv",
    sep="\t"
).sample(10_000, random_state=42)

s2 = pd.read_csv(
    PROJECT_ROOT / "dataset/train/train_source2.tsv",
    sep="\t",
    nrows=200_000
)

s3 = pd.read_csv(
    PROJECT_ROOT / "dataset/train/train_source3.tsv",
    sep="\t",
    nrows=200_000
)

print("S1:", s1.shape)
print("S2:", s2.shape)
print("S3:", s3.shape)


# ============================================================
# 2. LIGHTWEIGHT NORMALIZATION
# ============================================================

def basic_normalize(series):
    return (
        series.fillna("")
        .astype(str)
        .str.lower()
        .str.normalize("NFKC")
        .str.replace(r"[^\w\s]", " ", regex=True)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )


def prepare(df):

    df = df.copy()

    # Basic normalized fields
    df["name_norm"] = basic_normalize(
        df["business_name"]
    )

    df["address_norm"] = basic_normalize(
        df["business_address"]
    )

    # Token-sorted name
    df["name_sorted"] = df["name_norm"].apply(
        lambda x: " ".join(sorted(x.split()))
        if x else ""
    )

    # Address number
    df["address_number"] = (
        df["address_norm"]
        .str.extract(r"(\d+)", expand=False)
        .fillna("")
    )

    # Country
    df["country"] = (
        df["country"]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.strip()
    )

    return df


print("\nPreparing data...")

s1 = prepare(s1)
s2 = prepare(s2)
s3 = prepare(s3)

print("Preparation complete.")


# ============================================================
# 3. CREATE BLOCKING KEYS
# ============================================================

def make_block_keys(df):

    df = df.copy()

    # -----------------------------------------
    # BLOCK 1
    # Country + exact normalized name
    # -----------------------------------------

    df["block_name"] = (
        df["country"]
        + "|"
        + df["name_norm"]
    )

    # -----------------------------------------
    # BLOCK 2
    # Country + token-sorted name
    # -----------------------------------------

    df["block_name_sorted"] = (
        df["country"]
        + "|"
        + df["name_sorted"]
    )

    # -----------------------------------------
    # BLOCK 3
    # Country + first address number
    # -----------------------------------------

    df["block_number"] = (
        df["country"]
        + "|"
        + df["address_number"]
    )

    return df


s1 = make_block_keys(s1)
s2 = make_block_keys(s2)
s3 = make_block_keys(s3)


# ============================================================
# 4. BLOCKING FUNCTION
# ============================================================

def generate_block_candidates(s1, source):

    results = []

    # -----------------------------------------
    # BLOCK 1: exact normalized name
    # -----------------------------------------

    left = s1[
        s1["name_norm"] != ""
    ][
        ["entity_id", "block_name"]
    ]

    right = source[
        source["name_norm"] != ""
    ][
        ["entity_id", "block_name"]
    ]

    x = left.merge(
        right,
        on="block_name"
    )

    x = x[
        ["entity_id_x", "entity_id_y"]
    ]

    results.append(x)

    print("  Name block:", len(x))


    # -----------------------------------------
    # BLOCK 2: token-sorted name
    # -----------------------------------------

    left = s1[
        s1["name_sorted"] != ""
    ][
        ["entity_id", "block_name_sorted"]
    ]

    right = source[
        source["name_sorted"] != ""
    ][
        ["entity_id", "block_name_sorted"]
    ]

    x = left.merge(
        right,
        on="block_name_sorted"
    )

    x = x[
        ["entity_id_x", "entity_id_y"]
    ]

    results.append(x)

    print("  Sorted-name block:", len(x))


    # -----------------------------------------
    # BLOCK 3: country + address number
    # -----------------------------------------

    # NOTE:
    # This block is intentionally included only
    # to SEE how large it becomes.
    # We probably will NOT keep it alone.

    left = s1[
        s1["address_number"] != ""
    ][
        ["entity_id", "block_number"]
    ]

    right = source[
        source["address_number"] != ""
    ][
        ["entity_id", "block_number"]
    ]

    x = left.merge(
        right,
        on="block_number"
    )

    x = x[
        ["entity_id_x", "entity_id_y"]
    ]

    results.append(x)

    print("  Number block:", len(x))


    # -----------------------------------------
    # UNION
    # -----------------------------------------

    candidates = pd.concat(
        results,
        ignore_index=True
    )

    candidates.columns = [
        "source1_entity_id",
        "candidate_entity_id"
    ]

    candidates = candidates.drop_duplicates()

    return candidates


# ============================================================
# 5. RUN S1 -> S2
# ============================================================

print("\n==============================")
print("S1 -> S2 BLOCKING")
print("==============================")

candidates_s2 = generate_block_candidates(
    s1,
    s2
)

print(
    "\nTotal unique S2 candidates:",
    len(candidates_s2)
)


# ============================================================
# 6. RUN S1 -> S3
# ============================================================

print("\n==============================")
print("S1 -> S3 BLOCKING")
print("==============================")

candidates_s3 = generate_block_candidates(
    s1,
    s3
)

print(
    "\nTotal unique S3 candidates:",
    len(candidates_s3)
)


# ============================================================
# 7. COMBINE
# ============================================================

candidates = pd.concat(
    [
        candidates_s2,
        candidates_s3
    ],
    ignore_index=True
).drop_duplicates()


# ============================================================
# 8. FINAL SUMMARY
# ============================================================

print("\n==============================")
print("BLOCKING SUMMARY")
print("==============================")

print("S1 records:", len(s1))
print("S2 records:", len(s2))
print("S3 records:", len(s3))

print("Total candidate pairs:", len(candidates))

print(
    "Average candidates per S1:",
    round(len(candidates) / len(s1), 2)
)


# ============================================================
# 9. SEE ACTUAL CANDIDATES
# ============================================================

print("\n==============================")
print("SAMPLE CANDIDATES")
print("==============================")

display(
    candidates.head(30)
)
```

```python
# COUNTRY CONSISTENCY CHECK

def norm_country(x):
    if pd.isna(x):
        return ""
    return str(x).strip().lower()

# Normalize country
s1["country_norm"] = s1["country"].map(norm_country)
s2["country_norm"] = s2["country"].map(norm_country)
s3["country_norm"] = s3["country"].map(norm_country)

# Fast ID -> country lookups
s1_country = dict(zip(s1["entity_id"], s1["country_norm"]))
s2_country = dict(zip(s2["entity_id"], s2["country_norm"]))
s3_country = dict(zip(s3["entity_id"], s3["country_norm"]))

same_country = 0
different_country = 0
missing_country = 0
total_matches = 0

for _, row in gt.iterrows():

    source1_id = row["source1_entity_id"]
    source1_c = s1_country.get(source1_id, "")

    matched_ids = str(row["matched_entity_ids"]).split(",")

    for matched_id in matched_ids:
        matched_id = matched_id.strip()

        if not matched_id:
            continue

        total_matches += 1

        if matched_id in s2_country:
            target_c = s2_country[matched_id]
        elif matched_id in s3_country:
            target_c = s3_country[matched_id]
        else:
            continue

        if not source1_c or not target_c:
            missing_country += 1
        elif source1_c == target_c:
            same_country += 1
        else:
            different_country += 1

print("Total ground-truth matches :", total_matches)
print("Same country               :", same_country)
print("Different country          :", different_country)
print("Missing country            :", missing_country)

known = same_country + different_country

if known:
    print()
    print("Same-country rate          :", round(same_country / known, 6))
    print("Different-country rate     :", round(different_country / known, 6))
```

```python
# ============================================================
# BLOCKING RECALL CHECK
# ============================================================

# Ground truth for the sampled S1 records
gt_sample = gt[gt["source1_entity_id"].isin(S1["entity_id"])].copy()

# Map S1 entity_id -> sample row index
s1_id_to_idx = {
    entity_id: i
    for i, entity_id in enumerate(S1["entity_id"])
}

# Map S2/S3 entity_id -> sample row index
s2_id_to_idx = {
    entity_id: i
    for i, entity_id in enumerate(S2["entity_id"])
}

s3_id_to_idx = {
    entity_id: i
    for i, entity_id in enumerate(S3["entity_id"])
}

# Convert candidate sets to lookup sets
candidate_s2_set = candidates_s2
candidate_s3_set = candidates_s3

total_true = 0
retrieved_true = 0
missed_true = 0

missed_examples = []

for _, row in gt_sample.iterrows():

    s1_id = row["source1_entity_id"]

    if s1_id not in s1_id_to_idx:
        continue

    s1_idx = s1_id_to_idx[s1_id]

    matched_ids = str(row["matched_entity_ids"]).split(",")

    for matched_id in matched_ids:

        matched_id = matched_id.strip()

        if not matched_id:
            continue

        total_true += 1

        found = False

        # Is this true match in S2 sample candidates?
        if matched_id in s2_id_to_idx:
            s2_idx = s2_id_to_idx[matched_id]

            if (s1_idx, s2_idx) in candidate_s2_set:
                found = True

        # Is this true match in S3 sample candidates?
        elif matched_id in s3_id_to_idx:
            s3_idx = s3_id_to_idx[matched_id]

            if (s1_idx, s3_idx) in candidate_s3_set:
                found = True

        if found:
            retrieved_true += 1
        else:
            missed_true += 1

            if len(missed_examples) < 10:
                missed_examples.append(
                    (s1_id, matched_id)
                )


# ============================================================
# RESULTS
# ============================================================

print("========== BLOCKING RECALL ==========")

print("True matches in sample :", total_true)
print("Retrieved by blocking :", retrieved_true)
print("Missed by blocking    :", missed_true)

if total_true > 0:
    recall = retrieved_true / total_true
    print("\nBlocking recall:", round(recall, 6))
    print("Blocking recall %:", round(recall * 100, 4), "%")

print("\nFirst missed examples:")
for x in missed_examples:
    print(x)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors
import numpy as np

# ============================================================
# FUZZY NAME BLOCKING TEST
# ============================================================

# Use the normalized names we already created
s1_names = S1["name_norm"].fillna("")
s2_names = S2["name_norm"].fillna("")
s3_names = S3["name_norm"].fillna("")

# Character n-gram TF-IDF
vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(3, 5),
    min_df=2,
    max_features=200_000
)

# Fit on target records
s2_matrix = vectorizer.fit_transform(s2_names)
s1_matrix = vectorizer.transform(s1_names)

print("S2 TF-IDF shape:", s2_matrix.shape)

# Retrieve top K similar names
K = 20

nn_s2 = NearestNeighbors(
    n_neighbors=K,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1
)

nn_s2.fit(s2_matrix)

distances_s2, indices_s2 = nn_s2.kneighbors(s1_matrix)

# Convert to candidate pairs
fuzzy_candidates_s2 = set()

for i in range(len(S1)):
    for j in indices_s2[i]:
        fuzzy_candidates_s2.add((i, j))


# ============================================================
# S3
# ============================================================

s3_matrix = vectorizer.transform(s3_names)

nn_s3 = NearestNeighbors(
    n_neighbors=K,
    metric="cosine",
    algorithm="brute",
    n_jobs=-1
)

nn_s3.fit(s3_matrix)

distances_s3, indices_s3 = nn_s3.kneighbors(s1_matrix)

fuzzy_candidates_s3 = set()

for i in range(len(S1)):
    for j in indices_s3[i]:
        fuzzy_candidates_s3.add((i, j))


# ============================================================
# RESULTS
# ============================================================

print("\n========== FUZZY BLOCKING ==========")

print("S1 → S2 fuzzy candidates:", len(fuzzy_candidates_s2))
print("S1 → S3 fuzzy candidates:", len(fuzzy_candidates_s3))

total = len(fuzzy_candidates_s2) + len(fuzzy_candidates_s3)

print("Combined candidates:", total)
print("Average candidates per S1:", round(total / len(S1), 2))

print("\nSample fuzzy candidates:")

for i, j in list(fuzzy_candidates_s2)[:10]:
    print(
        S1.iloc[i]["business_name"],
        " <--> ",
        S2.iloc[j]["business_name"]
    )
```

```python
# ============================================================
# COUNTRY-AWARE FUZZY NAME BLOCKING
# ============================================================

K = 50

fuzzy_candidates_s2_country = set()
fuzzy_candidates_s3_country = set()

countries = S1["country_norm"].unique()

print("Countries:", countries)


# ============================================================
# S1 -> S2
# ============================================================

for country in countries:

    s1_idx = np.where(
        S1["country_norm"].values == country
    )[0]

    s2_idx = np.where(
        S2["country_norm"].values == country
    )[0]

    if len(s1_idx) == 0 or len(s2_idx) == 0:
        continue

    print(
        f"S1 -> S2 | {country} | "
        f"S1={len(s1_idx)} S2={len(s2_idx)}"
    )

    matrix_s1 = s1_matrix[s1_idx]
    matrix_s2 = s2_matrix[s2_idx]

    k_use = min(K, len(s2_idx))

    nn = NearestNeighbors(
        n_neighbors=k_use,
        metric="cosine",
        algorithm="brute",
        n_jobs=-1
    )

    nn.fit(matrix_s2)

    _, local_indices = nn.kneighbors(matrix_s1)

    for q, neighbors in enumerate(local_indices):

        original_s1_idx = s1_idx[q]

        for local_j in neighbors:
            original_s2_idx = s2_idx[local_j]

            fuzzy_candidates_s2_country.add(
                (original_s1_idx, original_s2_idx)
            )


# ============================================================
# S1 -> S3
# ============================================================

for country in countries:

    s1_idx = np.where(
        S1["country_norm"].values == country
    )[0]

    s3_idx = np.where(
        S3["country_norm"].values == country
    )[0]

    if len(s1_idx) == 0 or len(s3_idx) == 0:
        continue

    print(
        f"S1 -> S3 | {country} | "
        f"S1={len(s1_idx)} S3={len(s3_idx)}"
    )

    matrix_s1 = s1_matrix[s1_idx]
    matrix_s3 = s3_matrix[s3_idx]

    k_use = min(K, len(s3_idx))

    nn = NearestNeighbors(
        n_neighbors=k_use,
        metric="cosine",
        algorithm="brute",
        n_jobs=-1
    )

    nn.fit(matrix_s3)

    _, local_indices = nn.kneighbors(matrix_s1)

    for q, neighbors in enumerate(local_indices):

        original_s1_idx = s1_idx[q]

        for local_j in neighbors:
            original_s3_idx = s3_idx[local_j]

            fuzzy_candidates_s3_country.add(
                (original_s1_idx, original_s3_idx)
            )


# ============================================================
# RESULTS
# ============================================================

print("\n========== COUNTRY-AWARE FUZZY BLOCKING ==========")

print(
    "S1 -> S2 candidates:",
    len(fuzzy_candidates_s2_country)
)

print(
    "S1 -> S3 candidates:",
    len(fuzzy_candidates_s3_country)
)

total = (
    len(fuzzy_candidates_s2_country)
    + len(fuzzy_candidates_s3_country)
)

print("Combined candidates:", total)

print(
    "Average candidates per S1:",
    round(total / len(S1), 2)
)


# ============================================================
# SAMPLE
# ============================================================

print("\nSample S1 -> S2 candidates:")

for i, j in list(fuzzy_candidates_s2_country)[:15]:

    print(
        S1.iloc[i]["business_name"],
        " <--> ",
        S2.iloc[j]["business_name"]
    )
```

```python
import pandas as pd
import numpy as np
import re
from pathlib import Path

# ============================================================
# 1. LOAD FULL TRAIN DATA
# ============================================================

train_dir = Path.cwd().parent / "dataset" / "train"

S1 = pd.read_csv(
    train_dir / "train_source1.tsv",
    sep="\t"
)

S2 = pd.read_csv(
    train_dir / "train_source2.tsv",
    sep="\t"
)

S3 = pd.read_csv(
    train_dir / "train_source3.tsv",
    sep="\t"
)

GT = pd.read_csv(
    train_dir / "train_ground_truth.tsv",
    sep="\t"
)

print("Loaded:")
print("S1:", S1.shape)
print("S2:", S2.shape)
print("S3:", S3.shape)
print("GT:", GT.shape)


# ============================================================
# 2. LIGHT NORMALIZATION
# ============================================================

def norm(x):
    if pd.isna(x):
        return ""

    x = str(x).lower()
    x = re.sub(r"[^a-z0-9]+", " ", x)
    x = re.sub(r"\s+", " ", x)

    return x.strip()


def sorted_name(x):
    tokens = norm(x).split()

    # remove extremely weak legal tokens for this blocking key
    legal = {
        "inc", "incorporated", "llc", "ltd",
        "limited", "corp", "corporation",
        "company", "co", "plc", "pvt",
        "private"
    }

    tokens = [t for t in tokens if t not in legal]

    return " ".join(sorted(tokens))


def address_key(x):
    x = norm(x)

    # Keep numeric information because it is useful for addresses
    nums = re.findall(r"\d+", x)

    if not nums:
        return ""

    # First number + normalized address
    return nums[0] + "|" + x


for df in [S1, S2, S3]:

    df["country_norm"] = df["country"].map(norm)

    df["name_norm"] = df["business_name"].map(norm)

    df["name_sorted"] = df["business_name"].map(sorted_name)

    df["address_norm"] = df["business_address"].map(norm)

    df["address_key"] = df["business_address"].map(address_key)


# ============================================================
# 3. BUILD MULTI-PASS BLOCKING KEYS
#
# IMPORTANT:
# Every pass is country-aware.
# Candidates from ALL passes are UNIONED.
# ============================================================

def make_key(country, value):

    if not country or not value:
        return ""

    return country + "|" + value


for df in [S1, S2, S3]:

    df["key_name"] = [
        make_key(c, n)
        for c, n in zip(
            df["country_norm"],
            df["name_norm"]
        )
    ]

    df["key_sorted_name"] = [
        make_key(c, n)
        for c, n in zip(
            df["country_norm"],
            df["name_sorted"]
        )
    ]

    df["key_address"] = [
        make_key(c, a)
        for c, a in zip(
            df["country_norm"],
            df["address_norm"]
        )
    ]

    df["key_address_number"] = [
        make_key(c, a)
        for c, a in zip(
            df["country_norm"],
            df["address_key"]
        )
    ]


# ============================================================
# 4. CREATE CANDIDATE PAIRS USING PANDAS MERGES
# ============================================================

def blocking_pairs(source1, target, source_name):

    all_pairs = []

    # ----------------------------------------
    # PASS 1: EXACT NORMALIZED NAME
    # ----------------------------------------

    a = source1[
        ["entity_id", "key_name"]
    ].copy()

    b = target[
        ["entity_id", "key_name"]
    ].copy()

    b = b[b["key_name"] != ""]

    pairs = a.merge(
        b,
        on="key_name",
        how="inner",
        suffixes=("_s1", f"_{source_name}")
    )

    pairs = pairs[
        ["entity_id_s1", f"entity_id_{source_name}"]
    ]

    all_pairs.append(pairs)

    print(
        source_name,
        "PASS 1 exact name:",
        len(pairs)
    )


    # ----------------------------------------
    # PASS 2: SORTED TOKEN NAME
    # ----------------------------------------

    a = source1[
        ["entity_id", "key_sorted_name"]
    ].copy()

    b = target[
        ["entity_id", "key_sorted_name"]
    ].copy()

    b = b[b["key_sorted_name"] != ""]

    pairs = a.merge(
        b,
        on="key_sorted_name",
        how="inner",
        suffixes=("_s1", f"_{source_name}")
    )

    pairs = pairs[
        ["entity_id_s1", f"entity_id_{source_name}"]
    ]

    all_pairs.append(pairs)

    print(
        source_name,
        "PASS 2 sorted name:",
        len(pairs)
    )


    # ----------------------------------------
    # PASS 3: EXACT ADDRESS
    # ----------------------------------------

    a = source1[
        ["entity_id", "key_address"]
    ].copy()

    b = target[
        ["entity_id", "key_address"]
    ].copy()

    b = b[b["key_address"] != ""]

    pairs = a.merge(
        b,
        on="key_address",
        how="inner",
        suffixes=("_s1", f"_{source_name}")
    )

    pairs = pairs[
        ["entity_id_s1", f"entity_id_{source_name}"]
    ]

    all_pairs.append(pairs)

    print(
        source_name,
        "PASS 3 exact address:",
        len(pairs)
    )


    # ----------------------------------------
    # PASS 4: COUNTRY + ADDRESS NUMBER
    #
    # This can be a large pass, so we inspect
    # its size before deciding whether to keep it.
    # ----------------------------------------

    a = source1[
        ["entity_id", "key_address_number"]
    ].copy()

    b = target[
        ["entity_id", "key_address_number"]
    ].copy()

    b = b[b["key_address_number"] != ""]

    # Count target bucket sizes first
    sizes = b["key_address_number"].value_counts()

    # Do NOT allow huge buckets.
    # This prevents common numbers like "1" from
    # exploding candidate generation.
    valid_keys = sizes[
        sizes <= 500
    ].index

    a = a[a["key_address_number"].isin(valid_keys)]
    b = b[b["key_address_number"].isin(valid_keys)]

    pairs = a.merge(
        b,
        on="key_address_number",
        how="inner",
        suffixes=("_s1", f"_{source_name}")
    )

    pairs = pairs[
        ["entity_id_s1", f"entity_id_{source_name}"]
    ]

    all_pairs.append(pairs)

    print(
        source_name,
        "PASS 4 address-number:",
        len(pairs)
    )


    # ----------------------------------------
    # UNION ALL PASSES
    # ----------------------------------------

    result = pd.concat(
        all_pairs,
        ignore_index=True
    )

    result = result.drop_duplicates()

    result.columns = [
        "source1_entity_id",
        "matched_entity_id"
    ]

    print(
        source_name,
        "FINAL UNIQUE CANDIDATES:",
        len(result)
    )

    return result


# ============================================================
# 5. RUN BLOCKING S1 -> S2
# ============================================================

print("\n================ S1 -> S2 ================\n")

C2 = blocking_pairs(
    S1,
    S2,
    "s2"
)


# ============================================================
# 6. RUN BLOCKING S1 -> S3
# ============================================================

print("\n================ S1 -> S3 ================\n")

C3 = blocking_pairs(
    S1,
    S3,
    "s3"
)


# ============================================================
# 7. COMBINE
# ============================================================

C2["target_source"] = "S2"
C3["target_source"] = "S3"

C2 = C2.rename(
    columns={"matched_entity_id": "matched_entity_id"}
)

C3 = C3.rename(
    columns={"matched_entity_id": "matched_entity_id"}
)

CANDIDATES = pd.concat(
    [C2, C3],
    ignore_index=True
)

CANDIDATES = CANDIDATES.drop_duplicates(
    subset=[
        "source1_entity_id",
        "matched_entity_id"
    ]
)

print("\n============================================")
print("TOTAL CANDIDATE PAIRS:", len(CANDIDATES))
print("============================================")


# ============================================================
# 8. VALIDATE AGAINST FULL GROUND TRUTH
# ============================================================

# Convert GT into individual true pairs

true_pairs = []

for _, row in GT.iterrows():

    s1_id = row["source1_entity_id"]

    matched = str(
        row["matched_entity_ids"]
    ).split(",")

    for mid in matched:

        mid = mid.strip()

        if not mid:
            continue

        if mid == "nan":
            continue

        true_pairs.append(
            (s1_id, mid)
        )


TRUE = pd.DataFrame(
    true_pairs,
    columns=[
        "source1_entity_id",
        "matched_entity_id"
    ]
)

TRUE = TRUE.drop_duplicates()

print("\nTotal true pairs:", len(TRUE))


# ============================================================
# 9. SET-BASED RECALL
# ============================================================

candidate_set = set(
    zip(
        CANDIDATES["source1_entity_id"],
        CANDIDATES["matched_entity_id"]
    )
)

true_set = set(
    zip(
        TRUE["source1_entity_id"],
        TRUE["matched_entity_id"]
    )
)

retrieved = true_set & candidate_set
missed = true_set - candidate_set

recall = len(retrieved) / len(true_set)

print("\n============================================")
print("BLOCKING VALIDATION")
print("============================================")

print("True pairs       :", len(true_set))
print("Retrieved pairs  :", len(retrieved))
print("Missed pairs     :", len(missed))

print(
    "BLOCKING RECALL  :",
    round(recall, 6)
)

print(
    "BLOCKING RECALL %:",
    round(recall * 100, 4),
    "%"
)


# ============================================================
# 10. MISSED TRUE PAIRS
# ============================================================

MISSED = pd.DataFrame(
    list(missed),
    columns=[
        "source1_entity_id",
        "matched_entity_id"
    ]
)

print("\nFirst 20 missed true pairs:")

print(
    MISSED.head(20).to_string(index=False)
)


# ============================================================
# 11. SAVE CANDIDATES + MISSES
# ============================================================

output_dir = Path.cwd().parent / "outputs"
output_dir.mkdir(exist_ok=True)

CANDIDATES.to_csv(
    output_dir / "candidate_pairs_blocking_test.tsv",
    sep="\t",
    index=False
)

MISSED.to_csv(
    output_dir / "missed_true_pairs_blocking.tsv",
    sep="\t",
    index=False
)

print("\nSaved:")
print(
    output_dir / "candidate_pairs_blocking_test.tsv"
)
print(
    output_dir / "missed_true_pairs_blocking.tsv"
)
```

```python
# ============================================================
# MEMORY-SAFE EXACT BLOCKING RECALL
# DO NOT RERUN THE BLOCKING CELL
# ============================================================

import numpy as np
import pandas as pd

print("Building compact ground-truth keys...")

# ------------------------------------------------------------
# IMPORTANT:
# We already know from the previous run:
# Total actual GT pairs = 7,638,365
# ------------------------------------------------------------

MAX_GT_PAIRS = 7_638_365

gt_s2 = np.empty(MAX_GT_PAIRS, dtype=np.uint64)
gt_s3 = np.empty(MAX_GT_PAIRS, dtype=np.uint64)

n_s2 = 0
n_s3 = 0

# ------------------------------------------------------------
# Convert GT IDs into compact uint64 keys
#
# key = (S1 numeric ID << 32) | target numeric ID
#
# S2 and S3 are kept in separate arrays.
# ------------------------------------------------------------

for row in GT.itertuples(index=False):

    s1_id = str(row.source1_entity_id)

    # Extract numeric part of S1-xxxxxxxxx
    s1_num = int(s1_id.split("-", 1)[1])

    matched_ids = str(row.matched_entity_ids).split(",")

    for mid in matched_ids:

        mid = mid.strip()

        if not mid or mid == "nan":
            continue

        target_num = int(mid.split("-", 1)[1])

        key = (
            (np.uint64(s1_num) << np.uint64(32))
            | np.uint64(target_num)
        )

        if mid.startswith("S2-"):
            gt_s2[n_s2] = key
            n_s2 += 1

        elif mid.startswith("S3-"):
            gt_s3[n_s3] = key
            n_s3 += 1


# Trim unused space
gt_s2 = gt_s2[:n_s2]
gt_s3 = gt_s3[:n_s3]

# Sort for binary-search membership
gt_s2.sort()
gt_s3.sort()

print("Ground truth S2 pairs:", len(gt_s2))
print("Ground truth S3 pairs:", len(gt_s3))


# ============================================================
# VALIDATE CANDIDATES IN CHUNKS
# ============================================================

def validate_candidates(C, gt_keys, source_name, chunk_size=500_000):

    total_candidates = len(C)
    retrieved_true = 0

    print("\nValidating", source_name)
    print("Candidates:", total_candidates)

    for start in range(0, total_candidates, chunk_size):

        end = min(start + chunk_size, total_candidates)

        # Only take the two required columns
        chunk = C.iloc[
            start:end
        ]

        # Convert S1 IDs to numeric
        s1_nums = (
            chunk["source1_entity_id"]
            .str.split("-", n=1)
            .str[1]
            .astype(np.uint64)
            .to_numpy()
        )

        # Convert target IDs to numeric
        target_nums = (
            chunk["matched_entity_id"]
            .str.split("-", n=1)
            .str[1]
            .astype(np.uint64)
            .to_numpy()
        )

        # Compact pair key
        keys = (
            (s1_nums << np.uint64(32))
            | target_nums
        )

        # Exact membership using binary search
        positions = np.searchsorted(
            gt_keys,
            keys
        )

        valid = positions < len(gt_keys)

        found = np.zeros(
            len(keys),
            dtype=bool
        )

        found[valid] = (
            gt_keys[positions[valid]]
            == keys[valid]
        )

        retrieved_true += int(found.sum())

        if start % (chunk_size * 10) == 0:
            print(
                f"Processed {end:,} / {total_candidates:,}"
            )

    recall = retrieved_true / len(gt_keys)

    print(
        f"{source_name} retrieved true pairs:",
        retrieved_true
    )

    print(
        f"{source_name} recall:",
        round(recall * 100, 4),
        "%"
    )

    return retrieved_true


# ------------------------------------------------------------
# S2
# ------------------------------------------------------------

retrieved_s2 = validate_candidates(
    C2,
    gt_s2,
    "S2"
)


# ------------------------------------------------------------
# S3
# ------------------------------------------------------------

retrieved_s3 = validate_candidates(
    C3,
    gt_s3,
    "S3"
)


# ============================================================
# FINAL RESULT
# ============================================================

total_true = len(gt_s2) + len(gt_s3)
total_retrieved = retrieved_s2 + retrieved_s3
total_missed = total_true - total_retrieved

overall_recall = (
    total_retrieved / total_true
)

print("\n")
print("============================================")
print("        FINAL BLOCKING VALIDATION")
print("============================================")

print("Total true pairs      :", total_true)
print("Retrieved true pairs :", total_retrieved)
print("Missed true pairs     :", total_missed)

print(
    "BLOCKING RECALL       :",
    round(overall_recall * 100, 4),
    "%"
)

print("============================================")
```

```python
# ============================================================
# FINAL ENTITY RESOLUTION BLOCKER
# ============================================================
#
# Strategy:
#   1. Load S1 / S2 / S3 / Ground Truth
#   2. Lightweight normalization
#   3. Country-aware candidate retrieval
#   4. Character TF-IDF:
#        - business_name Top-30
#        - business_address Top-30
#   5. Union name + address candidates
#   6. Supplement with exact normalized blocks
#   7. Validate candidate recall against GT
#   8. Optionally write candidate_pairs.tsv incrementally
#
# IMPORTANT:
#   Start with RUN_MODE = "TEST"
#   Once TEST succeeds, change to "FULL"
#
# ============================================================

import os
import sys
import gc
import re
import subprocess
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
from scipy import sparse

warnings.filterwarnings("ignore")

# ============================================================
# CONFIG
# ============================================================

PROJECT_ROOT = Path(r"D:\ml-challenge")
DATA_DIR = PROJECT_ROOT / "dataset" / "train"
OUTPUT_DIR = PROJECT_ROOT / "outputs"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

S1_FILE = DATA_DIR / "train_source1.tsv"
S2_FILE = DATA_DIR / "train_source2.tsv"
S3_FILE = DATA_DIR / "train_source3.tsv"
GT_FILE = DATA_DIR / "train_ground_truth.tsv"

# ------------------------------------------------------------
# FIRST RUN:
# ------------------------------------------------------------
# "TEST" = 10k S1 + 200k target rows
#
# After this works and recall looks good:
# change to "FULL"
# ------------------------------------------------------------

RUN_MODE = "TEST"

N_TEST_S1 = 10_000
N_TEST_TARGET = 200_000

# Main retrieval parameter
K_NAME = 30
K_ADDRESS = 30

# Supplementary exact blocks
USE_EXACT_BLOCKS = True

# TF-IDF settings
NGRAM_RANGE = (3, 5)
MAX_FEATURES = 150_000
MIN_DF = 2

# Memory/performance
QUERY_CHUNK = 5_000
N_THREADS = max(1, (os.cpu_count() or 4) - 1)

# Output
CANDIDATE_FILE = OUTPUT_DIR / "candidate_pairs.tsv"

# ============================================================
# 1. INSTALL / IMPORT sparse_dot_topn
# ============================================================

try:
    from sparse_dot_topn import sp_matmul_topn
    print("sparse_dot_topn: available")
except ImportError:
    print("Installing sparse_dot_topn...")
    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "sparse_dot_topn",
        "-q"
    ])
    from sparse_dot_topn import sp_matmul_topn

from sklearn.feature_extraction.text import TfidfVectorizer


# ============================================================
# 2. BASIC NORMALIZATION
# ============================================================

def norm_text(s):
    """
    Conservative normalization for blocking.
    Does NOT perform aggressive transliteration.
    """
    if pd.isna(s):
        return ""

    s = str(s).lower()

    # Replace punctuation/symbols with spaces
    s = re.sub(r"[^0-9a-zA-Z\u00C0-\uFFFF]+", " ", s)

    # Collapse whitespace
    s = re.sub(r"\s+", " ", s).strip()

    return s


def normalize_df(df):
    df = df.copy()

    df["name_norm"] = (
        df["business_name"]
        .fillna("")
        .map(norm_text)
    )

    df["address_norm"] = (
        df["business_address"]
        .fillna("")
        .map(norm_text)
    )

    df["country_norm"] = (
        df["country"]
        .fillna("")
        .map(norm_text)
    )

    # Sorted-token name helps exact supplementary blocking
    def sort_tokens(x):
        if not x:
            return ""
        return " ".join(sorted(set(x.split())))

    df["name_sorted"] = df["name_norm"].map(sort_tokens)

    # Address number
    df["address_number"] = (
        df["address_norm"]
        .str.extract(r"\b(\d{1,6})\b", expand=False)
        .fillna("")
    )

    return df


# ============================================================
# 3. LOAD DATA
# ============================================================

print("\n" + "=" * 70)
print("LOADING DATA")
print("=" * 70)

S1 = pd.read_csv(S1_FILE, sep="\t", dtype=str)
S2 = pd.read_csv(S2_FILE, sep="\t", dtype=str)
S3 = pd.read_csv(S3_FILE, sep="\t", dtype=str)

GT = pd.read_csv(GT_FILE, sep="\t", dtype=str)

print("S1:", S1.shape)
print("S2:", S2.shape)
print("S3:", S3.shape)
print("GT:", GT.shape)

# Normalize
print("\nNormalizing...")

S1 = normalize_df(S1)
S2 = normalize_df(S2)
S3 = normalize_df(S3)

print("Normalization complete.")


# ============================================================
# 4. TEST/FULL SUBSETS
# ============================================================

if RUN_MODE.upper() == "TEST":

    B1 = S1.head(N_TEST_S1).copy()
    B2 = S2.head(N_TEST_TARGET).copy()
    B3 = S3.head(N_TEST_TARGET).copy()

    print("\nRUN MODE = TEST")
    print("B1:", B1.shape)
    print("B2:", B2.shape)
    print("B3:", B3.shape)

else:

    B1 = S1
    B2 = S2
    B3 = S3

    print("\nRUN MODE = FULL")
    print("B1:", B1.shape)
    print("B2:", B2.shape)
    print("B3:", B3.shape)


# ============================================================
# 5. GROUND TRUTH PAIRS
# ============================================================

print("\n" + "=" * 70)
print("PREPARING GROUND TRUTH")
print("=" * 70)

gt_pairs = []

for _, row in GT.iterrows():

    s1_id = row["source1_entity_id"]

    if pd.isna(row["matched_entity_ids"]):
        continue

    value = str(row["matched_entity_ids"]).strip()

    if not value:
        continue

    for target_id in value.split(","):

        target_id = target_id.strip()

        if target_id:
            gt_pairs.append((s1_id, target_id))

gt_pairs = pd.DataFrame(
    gt_pairs,
    columns=["s1_id", "target_id"]
)

print("Total GT pairs:", len(gt_pairs))

# ============================================================
# 6. SPARSE TF-IDF TOP-K FUNCTION
# ============================================================

def sparse_tfidf_topk(
    query_texts,
    target_texts,
    k=30,
    ngram_range=(3, 5),
    max_features=150_000,
    min_df=2,
    query_chunk=5000,
    n_threads=4
):
    """
    Memory-safe sparse TF-IDF retrieval.

    Returns:
        list of (query_local_index, target_local_index, similarity)
    """

    query_texts = (
        pd.Series(query_texts)
        .fillna("")
        .astype(str)
        .tolist()
    )

    target_texts = (
        pd.Series(target_texts)
        .fillna("")
        .astype(str)
        .tolist()
    )

    # --------------------------------------------------------
    # Fit vectorizer on TARGET corpus
    # --------------------------------------------------------

    vectorizer = TfidfVectorizer(
        analyzer="char",
        ngram_range=ngram_range,
        min_df=min_df,
        max_features=max_features,
        sublinear_tf=True,
        dtype=np.float32,
        norm="l2"
    )

    target_matrix = vectorizer.fit_transform(target_texts)

    print(
        f"      TF-IDF target matrix: "
        f"{target_matrix.shape}, "
        f"nnz={target_matrix.nnz:,}"
    )

    results = []

    # --------------------------------------------------------
    # Process queries in chunks
    # --------------------------------------------------------

    for start in range(0, len(query_texts), query_chunk):

        end = min(start + query_chunk, len(query_texts))

        q_matrix = vectorizer.transform(
            query_texts[start:end]
        )

        # A = queries
        # B = targets
        #
        # cosine similarity = normalized TF-IDF dot product
        #
        # sparse_dot_topn keeps only top-k results.
        C = sp_matmul_topn(
            q_matrix,
            target_matrix.T,
            top_n=k,
            threshold=0.0,
            sort=True,
            n_threads=n_threads
        )

        coo = C.tocoo()

        if coo.nnz:

            results.extend(
                zip(
                    coo.row.astype(np.int32) + start,
                    coo.col.astype(np.int32),
                    coo.data.astype(np.float32)
                )
            )

        del q_matrix
        del C
        del coo

        if (
            start == 0
            or end == len(query_texts)
            or start % (query_chunk * 10) == 0
        ):
            print(
                f"      queries {end:,}/{len(query_texts):,}"
            )

        gc.collect()

    del target_matrix
    del vectorizer
    gc.collect()

    return results


# ============================================================
# 7. EXACT BLOCKS
# ============================================================

def build_exact_maps(target):

    maps = {}

    # Exact normalized name
    maps["name"] = {}

    for i, value in enumerate(target["name_norm"].values):

        if value:
            maps["name"].setdefault(value, []).append(i)

    # Sorted name
    maps["name_sorted"] = {}

    for i, value in enumerate(target["name_sorted"].values):

        if value:
            maps["name_sorted"].setdefault(value, []).append(i)

    # Exact address
    maps["address"] = {}

    for i, value in enumerate(target["address_norm"].values):

        if value:
            maps["address"].setdefault(value, []).append(i)

    return maps


def exact_candidates(query_row, maps):

    out = set()

    name = query_row["name_norm"]
    sorted_name = query_row["name_sorted"]
    address = query_row["address_norm"]

    if name in maps["name"]:
        out.update(maps["name"][name])

    if sorted_name in maps["name_sorted"]:
        out.update(maps["name_sorted"][sorted_name])

    if address in maps["address"]:
        out.update(maps["address"][address])

    return out


# ============================================================
# 8. PROCESS ONE SOURCE
# ============================================================

def process_source(
    queries,
    targets,
    source_name
):

    print("\n" + "=" * 70)
    print(f"PROCESSING {source_name}")
    print("=" * 70)

    all_candidate_sets = [
        set() for _ in range(len(queries))
    ]

    # --------------------------------------------------------
    # COUNTRY PARTITION
    # --------------------------------------------------------

    query_countries = (
        queries["country_norm"]
        .fillna("")
        .astype(str)
        .values
    )

    target_countries = (
        targets["country_norm"]
        .fillna("")
        .astype(str)
        .values
    )

    unique_countries = sorted(
        set(query_countries) &
        set(target_countries)
    )

    print(
        "Countries processed:",
        len(unique_countries)
    )

    # --------------------------------------------------------
    # PROCESS EACH COUNTRY SEPARATELY
    # --------------------------------------------------------

    for country in unique_countries:

        q_idx = np.where(
            query_countries == country
        )[0]

        t_idx = np.where(
            target_countries == country
        )[0]

        if len(q_idx) == 0 or len(t_idx) == 0:
            continue

        print(
            f"\nCountry='{country}' "
            f"S1={len(q_idx):,} "
            f"{source_name}={len(t_idx):,}"
        )

        q_country = queries.iloc[q_idx]
        t_country = targets.iloc[t_idx]

        # ----------------------------------------------------
        # NAME TOP-K
        # ----------------------------------------------------

        print("   NAME TF-IDF...")

        name_results = sparse_tfidf_topk(
            q_country["name_norm"],
            t_country["name_norm"],
            k=K_NAME,
            ngram_range=NGRAM_RANGE,
            max_features=MAX_FEATURES,
            min_df=MIN_DF,
            query_chunk=QUERY_CHUNK,
            n_threads=N_THREADS
        )

        for q_local, t_local, score in name_results:

            q_global = q_idx[q_local]
            t_global = t_idx[t_local]

            all_candidate_sets[q_global].add(
                int(t_global)
            )

        del name_results
        gc.collect()

        # ----------------------------------------------------
        # ADDRESS TOP-K
        # ----------------------------------------------------

        print("   ADDRESS TF-IDF...")

        address_results = sparse_tfidf_topk(
            q_country["address_norm"],
            t_country["address_norm"],
            k=K_ADDRESS,
            ngram_range=NGRAM_RANGE,
            max_features=MAX_FEATURES,
            min_df=MIN_DF,
            query_chunk=QUERY_CHUNK,
            n_threads=N_THREADS
        )

        for q_local, t_local, score in address_results:

            q_global = q_idx[q_local]
            t_global = t_idx[t_local]

            all_candidate_sets[q_global].add(
                int(t_global)
            )

        del address_results
        gc.collect()

        # ----------------------------------------------------
        # EXACT BLOCKS
        # ----------------------------------------------------

        if USE_EXACT_BLOCKS:

            print("   EXACT BLOCKS...")

            maps = build_exact_maps(t_country)

            for q_global in q_idx:

                matches = exact_candidates(
                    queries.iloc[q_global],
                    maps
                )

                if matches:
                    for t_local in matches:

                        t_global = t_idx[t_local]

                        all_candidate_sets[q_global].add(
                            int(t_global)
                        )

            del maps
            gc.collect()

    # --------------------------------------------------------
    # CREATE CANDIDATE DATAFRAME
    # --------------------------------------------------------

    rows = []

    for q_idx, candidates in enumerate(
        all_candidate_sets
    ):

        for t_idx in candidates:

            rows.append(
                (
                    queries.iloc[q_idx]["entity_id"],
                    targets.iloc[t_idx]["entity_id"]
                )
            )

    result = pd.DataFrame(
        rows,
        columns=[
            "source1_entity_id",
            "matched_entity_id"
        ]
    )

    # Remove duplicates
    result = result.drop_duplicates(
        ignore_index=True
    )

    return result


# ============================================================
# 9. RUN S1 -> S2
# ============================================================

print("\n" + "=" * 70)
print("STARTING S1 -> S2")
print("=" * 70)

C2 = process_source(
    B1,
    B2,
    "S2"
)

print(
    "\nS1->S2 candidates:",
    f"{len(C2):,}"
)

print(
    "Average candidates/S1:",
    round(len(C2) / len(B1), 2)
)


# ============================================================
# 10. RUN S1 -> S3
# ============================================================

print("\n" + "=" * 70)
print("STARTING S1 -> S3")
print("=" * 70)

C3 = process_source(
    B1,
    B3,
    "S3"
)

print(
    "\nS1->S3 candidates:",
    f"{len(C3):,}"
)

print(
    "Average candidates/S1:",
    round(len(C3) / len(B1), 2)
)


# ============================================================
# 11. VALIDATE AGAINST GROUND TRUTH
# ============================================================

print("\n" + "=" * 70)
print("BLOCKING RECALL")
print("=" * 70)


def evaluate_recall(
    candidate_df,
    gt,
    s1_ids,
    target_ids,
    label
):

    s1_allowed = set(s1_ids)
    target_allowed = set(target_ids)

    # Only evaluate GT pairs that can possibly exist
    # in this benchmark subset.

    eligible = gt[
        gt["s1_id"].isin(s1_allowed)
        &
        gt["target_id"].isin(target_allowed)
    ].copy()

    candidate_set = set(
        zip(
            candidate_df["source1_entity_id"],
            candidate_df["matched_entity_id"]
        )
    )

    retrieved = sum(
        (
            row.s1_id,
            row.target_id
        ) in candidate_set
        for row in eligible.itertuples()
    )

    total = len(eligible)

    recall = (
        retrieved / total
        if total
        else np.nan
    )

    print(f"\n{label}")
    print("-" * 50)
    print("Eligible GT pairs :", f"{total:,}")
    print("Retrieved          :", f"{retrieved:,}")
    print("Missed             :", f"{total - retrieved:,}")
    print(
        "Recall              :",
        f"{recall:.4%}" if total else "N/A"
    )

    return {
        "label": label,
        "eligible": total,
        "retrieved": retrieved,
        "missed": total - retrieved,
        "recall": recall
    }


r2 = evaluate_recall(
    C2,
    gt_pairs,
    B1["entity_id"],
    B2["entity_id"],
    "S1 -> S2"
)

r3 = evaluate_recall(
    C3,
    gt_pairs,
    B1["entity_id"],
    B3["entity_id"],
    "S1 -> S3"
)


# ============================================================
# 12. OVERALL RECALL
# ============================================================

gt2 = gt_pairs[
    gt_pairs["s1_id"].isin(B1["entity_id"])
    &
    gt_pairs["target_id"].isin(B2["entity_id"])
]

gt3 = gt_pairs[
    gt_pairs["s1_id"].isin(B1["entity_id"])
    &
    gt_pairs["target_id"].isin(B3["entity_id"])
]

all_gt = set(
    zip(gt2["s1_id"], gt2["target_id"])
) | set(
    zip(gt3["s1_id"], gt3["target_id"])
)

all_candidates = set(
    zip(
        C2["source1_entity_id"],
        C2["matched_entity_id"]
    )
) | set(
    zip(
        C3["source1_entity_id"],
        C3["matched_entity_id"]
    )
)

retrieved_all = len(
    all_gt & all_candidates
)

overall_recall = (
    retrieved_all / len(all_gt)
    if all_gt
    else np.nan
)

print("\n" + "=" * 70)
print("OVERALL")
print("=" * 70)

print(
    "Eligible GT:",
    f"{len(all_gt):,}"
)

print(
    "Retrieved:",
    f"{retrieved_all:,}"
)

print(
    "Missed:",
    f"{len(all_gt) - retrieved_all:,}"
)

print(
    "Overall recall:",
    f"{overall_recall:.4%}"
)


# ============================================================
# 13. SAVE TEST CANDIDATES
# ============================================================

C2_OUT = OUTPUT_DIR / "candidate_pairs_S2_TEST.tsv"
C3_OUT = OUTPUT_DIR / "candidate_pairs_S3_TEST.tsv"

C2.to_csv(
    C2_OUT,
    sep="\t",
    index=False
)

C3.to_csv(
    C3_OUT,
    sep="\t",
    index=False
)

print("\nSaved:")
print(C2_OUT)
print(C3_OUT)


# ============================================================
# 14. FINAL DIAGNOSTIC
# ============================================================

avg2 = len(C2) / len(B1)
avg3 = len(C3) / len(B1)

print("\n" + "=" * 70)
print("FINAL BLOCKER DIAGNOSTIC")
print("=" * 70)

print(
    f"S2 candidates: {len(C2):,} "
    f"({avg2:.2f}/S1)"
)

print(
    f"S3 candidates: {len(C3):,} "
    f"({avg3:.2f}/S1)"
)

print(
    f"S2 recall: {r2['recall']:.4%}"
)

print(
    f"S3 recall: {r3['recall']:.4%}"
)

print(
    f"Overall recall: {overall_recall:.4%}"
)

print("\nDONE.")
```

```python
# ============================================================
# INSPECT BLOCKING MISSES FROM SAVED OUTPUTS
# ============================================================

from pathlib import Path
import pandas as pd

# ------------------------------------------------------------
# Find project automatically
# ------------------------------------------------------------

HERE = Path.cwd().resolve()

PROJECT_ROOT = None

for root in [
    HERE,
    HERE.parent,
    HERE.parent.parent,
    HERE.parent.parent.parent
]:
    if (
        (root / "dataset" / "train").exists()
        and
        (root / "outputs").exists()
    ):
        PROJECT_ROOT = root
        break

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Could not find project root."
    )

DATA_DIR = PROJECT_ROOT / "dataset" / "train"
OUTPUT_DIR = PROJECT_ROOT / "outputs"

print("Project:", PROJECT_ROOT)

# ------------------------------------------------------------
# Load original data
# ------------------------------------------------------------

S1 = pd.read_csv(
    DATA_DIR / "train_source1.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False
)

S2 = pd.read_csv(
    DATA_DIR / "train_source2.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False
)

S3 = pd.read_csv(
    DATA_DIR / "train_source3.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False
)

GT = pd.read_csv(
    DATA_DIR / "train_ground_truth.tsv",
    sep="\t",
    dtype=str,
    keep_default_na=False
)

# ------------------------------------------------------------
# Same TEST subsets used by blocker
# ------------------------------------------------------------

B1 = S1.head(10_000).copy()
B2 = S2.head(200_000).copy()
B3 = S3.head(200_000).copy()

# ------------------------------------------------------------
# Load candidate files
# ------------------------------------------------------------

C2 = pd.read_csv(
    OUTPUT_DIR / "candidate_pairs_S2_TEST.tsv",
    sep="\t",
    dtype=str
)

C3 = pd.read_csv(
    OUTPUT_DIR / "candidate_pairs_S3_TEST.tsv",
    sep="\t",
    dtype=str
)

print("\nLoaded candidates:")
print("C2:", C2.shape)
print("C3:", C3.shape)

# ------------------------------------------------------------
# Expand ground truth
# ------------------------------------------------------------

gt_rows = []

for row in GT.itertuples(index=False):

    s1_id = row.source1_entity_id
    matched = row.matched_entity_ids

    if not matched:
        continue

    for target_id in str(matched).split(","):

        target_id = target_id.strip()

        if target_id:
            gt_rows.append(
                (
                    s1_id,
                    target_id
                )
            )

GT_PAIRS = pd.DataFrame(
    gt_rows,
    columns=[
        "s1_id",
        "target_id"
    ]
).drop_duplicates()

print(
    "GT pairs:",
    f"{len(GT_PAIRS):,}"
)

# ------------------------------------------------------------
# Normalization for display
# ------------------------------------------------------------

def norm(x):
    return (
        str(x)
        .lower()
        .strip()
    )

# ------------------------------------------------------------
# Inspect misses
# ------------------------------------------------------------

def get_misses(
    candidates,
    query_df,
    target_df,
    label
):

    print("\n" + "=" * 80)
    print(label)
    print("=" * 80)

    query_ids = set(
        query_df["entity_id"]
    )

    target_ids = set(
        target_df["entity_id"]
    )

    eligible = GT_PAIRS[
        GT_PAIRS["s1_id"].isin(query_ids)
        &
        GT_PAIRS["target_id"].isin(target_ids)
    ].drop_duplicates()

    candidate_set = set(
        zip(
            candidates[
                "source1_entity_id"
            ],
            candidates[
                "matched_entity_id"
            ]
        )
    )

    missed = []

    for row in eligible.itertuples(index=False):

        pair = (
            row.s1_id,
            row.target_id
        )

        if pair not in candidate_set:
            missed.append(pair)

    print(
        "Eligible GT:",
        len(eligible)
    )

    print(
        "Missed:",
        len(missed)
    )

    if not missed:
        print("🎉 ZERO MISSES")
        return pd.DataFrame()

    # --------------------------------------------------------
    # Lookup source rows
    # --------------------------------------------------------

    s1_lookup = S1.set_index(
        "entity_id"
    )

    if target_df is B2:
        target_lookup = S2.set_index(
            "entity_id"
        )
    else:
        target_lookup = S3.set_index(
            "entity_id"
        )

    rows = []

    for s1_id, target_id in missed:

        a = s1_lookup.loc[s1_id]
        b = target_lookup.loc[target_id]

        rows.append({
            "s1_id": s1_id,
            "target_id": target_id,

            "S1_NAME":
                a["business_name"],

            "TARGET_NAME":
                b["business_name"],

            "S1_ADDRESS":
                a["business_address"],

            "TARGET_ADDRESS":
                b["business_address"],

            "S1_COUNTRY":
                a["country"],

            "TARGET_COUNTRY":
                b["country"],
        })

    result = pd.DataFrame(rows)

    pd.set_option(
        "display.max_columns",
        None
    )

    pd.set_option(
        "display.max_colwidth",
        120
    )

    display(result)

    return result


MISSES_S2 = get_misses(
    C2,
    B1,
    B2,
    "S1 -> S2 MISSES"
)

MISSES_S3 = get_misses(
    C3,
    B1,
    B3,
    "S1 -> S3 MISSES"
)
```
